```python
# pip install ipykernel -U --force-reinstall
"""load_data.py - load raw datasets and take a first look."""
from pathlib import Path
import numpy as np
import pandas as pd

# folder: week-02/data  (built from this file's location, so it works from anywhere)
DATA_DIR = Path(__file__).resolve().parent.parent / "data"


def load_faqs(filename="faq_raw.csv"):
    """Load the raw FAQ CSV exactly as it is (no cleaning here)."""
    return pd.read_csv(DATA_DIR / filename)


def load_courses(filename="courses.json"):
    """Load the courses JSON file into a DataFrame."""
    return pd.read_json(DATA_DIR / filename)


def first_look(df):
    """Run the 5 first-look commands from slide 8."""
    print("SHAPE:", df.shape)
    print("\nHEAD:\n", df.head())
    print("\nINFO:")
    df.info()
    print("\nDESCRIBE:\n", df.describe())
    print("\nBLANKS PER COLUMN:\n", df.isna().sum())


if __name__ == "__main__":
    print("===== FAQS =====")
    first_look(load_faqs())
    print("\n===== COURSES =====")
    print(load_courses())
```
